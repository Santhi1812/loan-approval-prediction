# Loan Approval Prediction using Machine Learning

Loans are a major requirement of the modern world. By this alone, banks receive a major portion of the total profit. It is beneficial for students to manage their education and living expenses, and for individuals to purchase various luxuries, such as houses and cars. But when it comes to deciding whether the applicant's profile is relevant to be granted a loan or not, banks have to look after many aspects.

Here, we will be using machine learning algorithms to ease their work and predict whether the candidate’s profile is relevant or not, using key features like Marital Status, Education, Applicant Income, Credit History, etc.

## Importing Libraries and Dataset

Firstly, we import the required libraries:

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

Now we load the dataset `data/LoanApprovalPrediction.csv`:

```python
data = pd.read_csv("../data/LoanApprovalPrediction.csv")
data.head(5)
```

## Data Preprocessing and Visualization

Get the number of columns of object datatype:

```python
obj = (data.dtypes == 'object')
print("Categorical variables:", len(list(obj[obj].index)))
```

As `Loan_ID` is completely unique and not correlated with any other columns, we drop it:

```python
# Dropping Loan_ID column
data.drop(['Loan_ID'], axis=1, inplace=True)
data.head(5)
```

Visualize all the unique values in categorical columns using a barplot. This shows which values dominate our dataset:

```python
obj = (data.dtypes == 'object')
object_cols = list(obj[obj].index)
plt.figure(figsize=(18, 36))
index = 1

for col in object_cols:
  y = data[col].value_counts()
  plt.subplot(11, 4, index)
  plt.xticks(rotation=90)
  sns.barplot(x=list(y.index), y=y)
  index += 1

plt.tight_layout()
plt.show()
```

As all the categorical values are binary or simple classes, we can use `LabelEncoder` to convert them into integer datatypes:

```python
# Import label encoder
from sklearn import preprocessing
  
# label_encoder object knows how to understand word labels
label_encoder = preprocessing.LabelEncoder()
obj = (data.dtypes == 'object')
for col in list(obj[obj].index):
  data[col] = label_encoder.fit_transform(data[col])
```

Verify the object datatype columns again to ensure they are all encoded:

```python
obj = (data.dtypes == 'object')
print("Categorical variables remaining:", len(list(obj[obj].index)))
```

Visualize feature correlations using a heatmap:

```python
plt.figure(figsize=(12, 6))
sns.heatmap(data.corr(), cmap='BrBG', fmt='.2f', linewidths=2, annot=True)
plt.show()
```

Use a catplot to visualize Gender and Marital Status against Loan Status:

```python
sns.catplot(x="Gender", y="Married", hue="Loan_Status", kind="bar", data=data)
plt.show()
```

Impute missing values using column means:

```python
for col in data.columns:
  data[col] = data[col].fillna(data[col].mean()) 
  
print("Missing values check:")
print(data.isna().sum())
```

## Splitting Dataset

```python
from sklearn.model_selection import train_test_split

X = data.drop(['Loan_Status'], axis=1)
Y = data['Loan_Status']
print("X shape:", X.shape, "Y shape:", Y.shape)

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.4, random_state=1)
print(f"X_train shape: {X_train.shape}, X_test shape: {X_test.shape}")
```

## Model Training and Evaluation

We train and evaluate KNeighborsClassifier, RandomForestClassifier, SVC, and LogisticRegression:

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn import metrics

knn = KNeighborsClassifier(n_neighbors=3)
rfc = RandomForestClassifier(n_estimators=7, criterion='entropy', random_state=7)
svc = SVC()
lc = LogisticRegression(max_iter=1000)

# Predictions on training set
for clf in (rfc, knn, svc, lc):
    clf.fit(X_train, Y_train)
    Y_pred = clf.predict(X_train)
    print("Accuracy score of ", 
          clf.__class__.__name__, 
          " = ", 100 * metrics.accuracy_score(Y_train, Y_pred))
```

Now we predict on the testing set to compare performance:

```python
# Predictions on testing set
for clf in (rfc, knn, svc, lc):
    clf.fit(X_train, Y_train)
    Y_pred = clf.predict(X_test)
    print("Accuracy score of ", 
          clf.__class__.__name__, 
          " = ", 100 * metrics.accuracy_score(Y_test, Y_pred))
```
